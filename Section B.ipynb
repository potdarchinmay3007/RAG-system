```python
#Section B — RAG Fundamentals
#B1. Fixed-Size Chunking with Overlap
#Question: Take a small corpus of 5–6 short paragraphs. Split it into fixed-size chunks with some overlap. Print the chunks and explain why you chose that chunk size and overlap.

text = """
Artificial Intelligence allows computers to perform tasks that normally
require human intelligence. Machine learning is a part of AI that learns
patterns from data.
RAG stands for Retrieval Augmented Generation. It retrieves relevant
information from a knowledge source before generating an answer.
Vector databases are used to store numerical representations called
embeddings. These embeddings help find similar pieces of information.
Chunking means breaking a large document into smaller pieces. It makes
documents easier to search and retrieve.
Dense retrieval converts text into embeddings and finds information based
on semantic similarity. """

#make chunks, embeddings from this data and search for it......
chunk_size = 100
overlap = 20
chunks = []
for i in range(0, len(text), chunk_size - overlap):
    chunks.append(text[i:i + chunk_size])

# This makes chunks of the data, and they overlap with each other by 20 characters cause we need  to connect the chuncks if there is an imp. sentence at the end will get divide that's why we need to overlap the chunks.
#so the next chunk will be of 80 charecters
for chunk in chunks:
    print(chunk)

```
